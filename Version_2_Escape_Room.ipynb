```python
# ==========================================
# AI ESCAPE ROOM - VERSION 2
# Improved Interactive Escape Room
# Developed using Python, Gemini AI and Gradio
# ==========================================

print("=" * 50)
print("       THE MYSTERY OF THE LOCKED LIBRARY")
print("              VERSION 2")
print("=" * 50)

print("\nWelcome to Version 2!")
print("Explore the library, solve puzzles and escape.")
print("\nImprovements in this version:")
print("1. Structured game state")
print("2. Inventory management")
print("3. Validated puzzle solutions")
print("4. Secure four-digit exit code")
print("5. Error handling")
print("6. Automated testing")
```

```python

# ==========================================
# STEP 2: GAME STATE AND INVENTORY SYSTEM
# ==========================================

# Store the complete game progress
game_state = {
    "player_name": "Player",
    "visited": [],
    "inventory": [],
    "clues": {},
    "drawer_unlocked": False,
    "exit_unlocked": False,
    "escaped": False,
    "attempts": 0
}

# Puzzle configuration
PUZZLES = {
    "desk_code": "24",
    "bookshelf_code": "19",
    "drawer_code": "2419",
    "exit_code": "7316"
}

# Available objects in the library
ROOM_OBJECTS = [
    "desk",
    "bookshelf",
    "painting",
    "drawer",
    "exit door"
]

# Function to display inventory
def show_inventory():
    print("\n========== PLAYER INVENTORY ==========")

    if game_state["inventory"]:
        for item in game_state["inventory"]:
            print("-", item)
    else:
        print("Your inventory is empty.")

    print("======================================")

# Function to display discovered clues
def show_clues():
    print("\n========== DISCOVERED CLUES ==========")

    if game_state["clues"]:
        for clue, value in game_state["clues"].items():
            print(f"{clue}: {value}")
    else:
        print("No clues discovered yet.")

    print("======================================")

print("Game state initialized successfully!")
print("Available objects:", ROOM_OBJECTS)
print("Inventory system ready!")
print("Puzzle system ready!")
```

```python

# ==========================================
# STEP 3: PUZZLE AND OBJECT INTERACTION
# ==========================================

def examine_object(object_name, code=None):

    object_name = object_name.lower().strip()

    # DESK PUZZLE
    if object_name == "desk":

        game_state["visited"].append("desk")

        game_state["clues"]["first_digits"] = PUZZLES["desk_code"]

        print("\nYou examine the old mahogany desk.")
        print("You discover a hidden note!")
        print("The note reveals the first two digits: 24")

    # BOOKSHELF PUZZLE
    elif object_name == "bookshelf":

        game_state["visited"].append("bookshelf")

        game_state["clues"]["second_digits"] = PUZZLES["bookshelf_code"]

        print("\nYou examine the mysterious bookshelf.")
        print("Behind an old book, you discover another clue!")
        print("The clue reveals the next two digits: 19")

    # PAINTING PUZZLE
    elif object_name == "painting":

        game_state["visited"].append("painting")

        if "Brass Key" not in game_state["inventory"]:
            game_state["inventory"].append("Brass Key")

        print("\nYou carefully inspect the painting.")
        print("Behind its frame, you discover a brass key!")
        print("Brass Key added to your inventory.")

    # DRAWER PUZZLE
    elif object_name == "drawer":

        if "Brass Key" not in game_state["inventory"]:
            print("\nThe drawer is locked.")
            print("You need to find the brass key first.")

        elif "first_digits" not in game_state["clues"] or "second_digits" not in game_state["clues"]:
            print("\nYou need to discover both number clues first.")

        elif game_state["drawer_unlocked"]:
            print("\nThe drawer is already open.")
            print("You have already collected the exit card.")

        elif code is None:
            print("\nThe drawer has a four-digit combination lock.")
            print("Enter the combination using:")
            print("Your two clues: 24 and 19")

        elif str(code).strip() == PUZZLES["drawer_code"]:

            game_state["drawer_unlocked"] = True

            if "Exit Card" not in game_state["inventory"]:
                game_state["inventory"].append("Exit Card")

            game_state["clues"]["final_digits"] = PUZZLES["exit_code"]

            print("\nSUCCESS! The drawer opens.")
            print("You discover a mysterious brass exit card.")
            print("The card reveals the final four-digit code: 7316")
            print("Exit Card added to your inventory.")

        else:
            game_state["attempts"] += 1
            print("\nIncorrect combination!")
            print("Try combining the two clues again.")

    # EXIT DOOR PUZZLE
    elif object_name in ["exit door", "door"]:

        if "Exit Card" not in game_state["inventory"]:
            print("\nThe exit door remains locked.")
            print("You must find the exit card first.")

        elif game_state["escaped"]:
            print("\nYou have already escaped the library!")

        elif code is None:
            print("\nThe keypad requests a four-digit code.")
            print("Enter the code from your exit card.")

        elif str(code).strip() == PUZZLES["exit_code"]:

            game_state["exit_unlocked"] = True
            game_state["escaped"] = True

            print("\n" + "=" * 45)
            print("CONGRATULATIONS!")
            print("YOU HAVE ESCAPED THE LOCKED LIBRARY!")
            print("=" * 45)

        else:
            game_state["attempts"] += 1
            print("\nIncorrect exit code!")
            print("The door remains locked.")

    else:
        print("\nSorry, that object does not exist.")
        print("Available objects:", ROOM_OBJECTS)
```
