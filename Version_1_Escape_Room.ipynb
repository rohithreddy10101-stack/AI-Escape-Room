```python

GAME_STORY = """
You are the Game Master of an AI Escape Room.

GAME NAME: The Mystery of the Locked Library

STORY:
The player wakes up inside a mysterious locked library.
The exit door is protected by a four-digit code.

OBJECTS IN THE ROOM:
1. Desk - contains a hidden note.
2. Bookshelf - contains hidden number clues.
3. Painting - hides a small key.
4. Locked drawer - requires the key and a code.
5. Exit door - requires a four-digit code.

PUZZLES:
Puzzle 1: Examine the desk to find the first two digits: 24.
Puzzle 2: Examine the bookshelf to find the last two digits: 19.
Puzzle 3: Examine the painting to discover a hidden key.

The player must:
- Find the desk clue.
- Find the bookshelf clue.
- Collect the key from the painting.
- Open the drawer using code 2419.
- Discover the final exit code 7316.
- Enter the correct exit code to escape.

GAME RULES:
- Respond naturally to player messages.
- Describe objects when examined.
- Give hints when requested.
- Remember previous player actions.
- Do not reveal all clues immediately.
- Do not claim the player has escaped without the correct code.

You are a friendly but mysterious Game Master.
"""

print("Game story created successfully!")
```

```python

conversation_history = []

print("Conversation history initialized!")
```

```python

def play_game_v1(player_message):

    conversation_history.append(
        "Player: " + player_message
    )

    prompt = GAME_STORY + "\n\n"

    prompt += "\n".join(conversation_history)

    prompt += "\nGame Master:"

    response = client.models.generate_content(
        model=MODEL_NAME,
        contents=prompt
    )

    ai_response = response.text

    conversation_history.append(
        "Game Master: " + ai_response
    )

    return ai_response


print("AI Game Master created successfully!")
```

```python

welcome = play_game_v1(
    "I have just woken up. Where am I?"
)

print(welcome)
```

```python

answer = play_game_v1(
    "I want to examine the bookshelf."
)

print(answer)
```

```python

conversation_history.clear()

print("Game conversation reset successfully!")
```

```python

def chat_with_game(message, history):

    if not message.strip():
        return history, ""

    # Send player's message to Gemini
    answer = play_game_v1(message)

    # Add messages to chat history
    history = history + [
        {"role": "user", "content": message},
        {"role": "assistant", "content": answer}
    ]

    return history, ""
```

```python

def restart_game():

    conversation_history.clear()

    return [], ""
```

```python

with gr.Blocks(title="AI Escape Room") as demo:

    gr.Markdown(
        """
        # THE MYSTERY OF THE LOCKED LIBRARY

        Welcome, player!

        Explore the mysterious library.
        Find clues, solve puzzles and escape!
        """
    )

    chatbot = gr.Chatbot(
        label="Game Master",
        height=450
    )

    message = gr.Textbox(
        label="Your Action",
        placeholder="Example: I want to examine the desk..."
    )

    with gr.Row():
        send_btn = gr.Button("Send", variant="primary")
        restart_btn = gr.Button("Restart Game")

    send_btn.click(
        chat_with_game,
        inputs=[message, chatbot],
        outputs=[chatbot, message]
    )

    message.submit(
        chat_with_game,
        inputs=[message, chatbot],
        outputs=[chatbot, message]
    )

    restart_btn.click(
        restart_game,
        outputs=[chatbot, message]
    )

print("Gradio interface created successfully!")
```

```python

def chat_with_game(message, history):

    if not message.strip():
        return history, ""

    answer = play_game_v1(message)

    history = history or []

    history = history + [
        (message, answer)
    ]

    return history, ""
```
